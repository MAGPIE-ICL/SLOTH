```python
import numpy as np
from scipy.integrate import quad
from scipy.stats import norm


def sigma_s2_from_Mach(M, b):
    """
    Variance of s = ln(rho/rho0) for isothermal supersonic turbulence:
        sigma_s^2 = ln(1 + b^2 M^2)
    """
    return np.log(1.0 + (b * M) ** 2)


def p_grad_component_z(x, sigma_grad_rho, sigma_s=None, M=None, b=None,
                       zmax=8.0, epsabs=1e-10, epsrel=1e-8):
    """
    Stable PDF for one density-gradient component using z = (s-mu_s)/sigma_s.
    """

    if sigma_grad_rho <= 0:
        raise ValueError("sigma_grad_rho must be > 0")

    if sigma_s is None:
        if M is None or b is None:
            raise ValueError("Provide either sigma_s, or both M and b.")
        sigma_s2 = np.log(1.0 + (b * M) ** 2)
        sigma_s = np.sqrt(sigma_s2)
    else:
        sigma_s2 = sigma_s**2

    mu_s = -0.5 * sigma_s2
    log_pref = sigma_s2 / 2.0 - np.log(np.sqrt(2.0 * np.pi) * sigma_grad_rho)

    def _single_val(xi):
        xi2 = float(xi) ** 2

        def integrand(z):
            s = mu_s + sigma_s * z

            # standard normal weight in z
            log_pz = -0.5 * z**2 - 0.5 * np.log(2.0 * np.pi)

            logA = np.log(xi2 + 1e-300) + sigma_s2 - np.log(2.0 * sigma_grad_rho**2) - 2.0 * s
            A = np.exp(np.clip(logA, -745, 700))

            logf = log_pref + log_pz - s - A
            return np.exp(np.clip(logf, -745, 700))

        val, _ = quad(integrand, -zmax, zmax, epsabs=epsabs, epsrel=epsrel, limit=200)
        return val

    x = np.asarray(x)
    if x.ndim == 0:
        return _single_val(float(x))
    return np.array([_single_val(xx) for xx in x])
```
